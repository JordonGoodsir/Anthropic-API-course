```python
from anthropic import Anthropic

client = Anthropic()
model = "claude-sonnet-4-6"
```

```python
def add_user_message(messages, text):
    user_message = {"role": "user", "content": text}
    messages.append(user_message)

def add_assistant_message(messages, text):
    assistant_message = {"role": "assistant", "content": text}
    messages.append(assistant_message)

def chat(messages, system=None):
    params = {
        "model": model,
        "max_tokens": 1000,
        "messages": messages,
    }
    
    if system:
        params["system"] = system
    
    message = client.messages.create(**params)
    return message.content[0].text
```

```python
messages = []

print("Terminal chat with Claude. Type 'exit' or 'quit' to end.\n")

system_prompt = """
You are an impatient math tutor with deep seeded hatred towards the student.
Do not directly answer a student's questions.
Guide them to a solution step by step begrudgingly.
"""

while True:
    user_input = input("You: ").strip()
    if user_input.lower() in {"exit", "quit"}:
        print("Goodbye!")
        break
    if not user_input:
        continue

    print(f"\nYou: {user_input}\n")

    add_user_message(messages, user_input)
    response = chat(messages, system=system_prompt)
    add_assistant_message(messages, response)
    print(f"Claude: {response}\n")
```
